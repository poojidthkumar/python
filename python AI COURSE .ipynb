# Assemble the game

```python
def kaprekar_step(number):
    digits = f"{number:04d}"

    descending = int("".join(sorted(digits, reverse=True)))
    ascending = int("".join(sorted(digits)))

    return descending - ascending



def play_kaprekar(number):

    if not 1000 <= number <= 9999:
        print("Please enter a four-digit number.")
        return

    if len(set(f"{number:04d}")) == 1:
        print("Choose a number with at least two different digits.")
        return

    moves = 0
    print("Starting number:", number)

    while number != 6174:
        number = kaprekar_step(number)
        moves += 1
        print(f"Move {moves}: {number:04d}")

    print("Congratulations! You reached 6174.")
    print("Total moves:", moves)

try:
    starting_number = int(input("Enter a four-digit number: "))
    play_kaprekar(starting_number)

except ValueError:
    print("Invalid input. Please enter digits only.")

```

# inheritance

```python
class Vehicle:
    def start(self):
        print("Vehical started....")

class car(Vehicle):
    def open_boot(self):
        print("Car boot opened...")

class bus(Vehicle):
    def carry_passenger(self):
        print("Bus is carrying passenger......")

Car = car()
Bus = bus()

Car.start()
Car.open_boot()

Bus.start()
Bus.carry_passenger()
```

# Python OOP – Easy Formulas
1. Class: Blueprint = Class
2. Object: Class + () = Object creation
3. Inheritance: Parent class + Child class = Inheritance
4. Polymorphism: Same method name + Different behaviour = Polymorphism
5. Encapsulation: Data + Methods combined in one class = Encapsulation
6. Abstraction: Hide implementation details + Show essential features = Abstraction

```python
class UPI:
    def pay(self,amount):
        print("paid using UPI:",amount)

class CreditCard:
    def pay(self,amount):
        print("paid using credit card:",amount)

class CashOnDelivery:
    def pay(self,amount):
        print("paid in cash on delivery:",amount)
        
payments = [UPI(),CreditCard(),CashOnDelivery()]

for method in payments:
    method.pay(1000)
```
